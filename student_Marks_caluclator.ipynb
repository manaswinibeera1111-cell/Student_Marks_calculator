```python
name=input("Enter student name")
print("student",name)
maths=int(input("enter maths marks:"))
physics=int(input("enter physics marks"))
chemistry=int(input("enter chemistry marks"))
total=maths+physics+chemistry
print(total)
average=total/3
print(average)
```

```python
if average>=90:
  grade="A"
elif average>=75:
  grade="B"
elif average>=60:
  grade="C"
elif average>=40:
  grade="D"
else:
  grade="Fail"
print("Grade:",grade)
```

```python
if(average>=40):
  status="Pass"
else:
  status="Fail"
print("Staus",status)
```

```python
print("======Student Result======")
print("Name:",name)
print("Maths:",maths)
print("physics:",physics)
print("chemistry:",chemistry)
print("Total:",total)
print("Average:",average)
print("Grade:",grade)
print("Status:",status)
print("======================")
```
